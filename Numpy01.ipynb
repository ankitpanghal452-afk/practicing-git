Indexing and Slicing 2D Arrays in NumPy

```python
import numpy as np
arr = np.arange(1, 26).reshape(5,5)
print(arr)
```

```python
print(arr[0,])
```

```python
print(arr[-1])
```

```python
print(arr[:,0])
```

```python
print(arr[:,-1])
```

```python
import numpy as np
arr = np.arange(1, 26).reshape(5,5)
print(arr)
```

```python
print([arr[1:3,1:3]])
```

```python
print(arr[2:,2:])
```

```python
print(arr[0::2])
```

```python
print(arr[:,0::2])
```

```python
print(arr[::-1])
```

```python
print(arr[:,::-1])
```

Indexing and Slicing 3D Arrays in NumPy

```python
arr3 = np.arange(1,25).reshape(2,3,4)
print(arr3)
```

```python
print(arr3[0])
```

```python
print(arr3[0,1])
```

```python
print(arr3[1,1])
```

```python
print(arr3[1,0,2])
```

```python
print(arr3[0,1::,1:-1])
```

```python
arr3 = np.arange(1,25).reshape(2,3,4)
print(arr3)
```

```python
print(arr3[1,:,1:-1])
```

```python
print(arr3[0:,:,-1])
```

```python
print(arr3[0:,0,])
```

```python
print(arr3[0:,1,1:3])
```
