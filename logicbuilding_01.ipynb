```python
def __init__(self,nums,k):
    self.nums=nums
    self.k=k
def result(self):
    self.nums=sorted(self.nums)
    left=0
    right=len(self.nums)-1
    while left<right:
        total=self.nums[left]+self.nums[right]
        if total==self.k:
            return [left,right]
        elif total<self.k:
            left+=1
        else:
            right-=1
    return -1
```
